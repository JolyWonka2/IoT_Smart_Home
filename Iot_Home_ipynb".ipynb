```python

# ============================================================
# РОЗДІЛ 3. РОЗРОБКА ПРОГРАМНОГО ЗАБЕЗПЕЧЕННЯ ДЛЯ ОБРОБКИ
# ТА АНАЛІЗУ ДАНИХ IOT З ВИКОРИСТАННЯМ EDGE ТА CLOUD COMPUTING
#
# Варіант 9. Енергозберігаюча система розумного будинку
#
# Середовище виконання: Google Colab
# Мова програмування: Python
# ============================================================


# ============================================================
# 1. ІМПОРТ БІБЛІОТЕК ТА ПОЧАТКОВЕ НАЛАШТУВАННЯ
# ============================================================

import os
import io
import json
import time
import sqlite3
import random
import gzip
import zlib
import bz2
import zipfile
import warnings

from datetime import datetime, timedelta

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

warnings.filterwarnings("ignore")

# Фіксуємо генератори випадкових чисел,
# щоб результати можна було відтворити повторно.
np.random.seed(42)
random.seed(42)

# Створюємо папки для результатів роботи.
os.makedirs("data", exist_ok=True)
os.makedirs("plots", exist_ok=True)
os.makedirs("reports", exist_ok=True)

print("Бібліотеки успішно підключено.")
print("Початок виконання програми.")


# ============================================================
# 2. ЕМУЛЮВАННЯ РОБОТИ СЕНСОРІВ РОЗУМНОГО БУДИНКУ
# ============================================================

# Для моделювання використовується чотири сенсори:
# 1 - освітленість (lux);
# 2 - споживання електроенергії (kWh);
# 3 - температура (°C);
# 4 - рух (0 або 1).

NUM_RECORDS = 2500
NUM_SENSORS = 4

# 2500 часових точок із кроком 5 хвилин.
# Кожна точка містить показники чотирьох сенсорів.
# Загальна кількість записів становить 10000.

START_TIME = datetime.utcnow().replace(microsecond=0)

timestamps = [
    START_TIME + timedelta(minutes=5 * i)
    for i in range(NUM_RECORDS)
]

hours = np.array([
    t.hour + t.minute / 60
    for t in timestamps
])

days = np.array([
    (t - START_TIME).total_seconds() / 86400
    for t in timestamps
])

# ------------------------------------------------------------
# 2.1. Моделювання активності мешканців
# ------------------------------------------------------------

# Імовірність присутності людей залежить від часу доби.
occupancy_probability = np.where(
    (hours >= 7) & (hours < 23),
    0.75,
    0.10
)

# Генеруємо показники датчика руху.
motion = (
    np.random.random(NUM_RECORDS) < occupancy_probability
).astype(int)

# ------------------------------------------------------------
# 2.2. Моделювання рівня освітленості
# ------------------------------------------------------------

# Природне освітлення залежить від часу доби.
daylight = np.maximum(
    0,
    np.sin((hours - 6) * np.pi / 12)
)

# Якщо в приміщенні є люди, може використовуватися штучне світло.
artificial_light = motion * np.random.uniform(
    100, 250, NUM_RECORDS
)

light = (
    daylight * 450
    + artificial_light
    + np.random.normal(0, 25, NUM_RECORDS)
)

light = np.clip(light, 0, 1000)

# ------------------------------------------------------------
# 2.3. Моделювання температури
# ------------------------------------------------------------

temperature = (
    21
    + 2 * np.sin((hours - 8) * np.pi / 12)
    + 0.5 * np.sin(days * 2 * np.pi)
    + np.random.normal(0, 0.4, NUM_RECORDS)
)

# ------------------------------------------------------------
# 2.4. Моделювання споживання електроенергії
# ------------------------------------------------------------

# Базове споживання електроенергії.
energy = (
    0.025
    + motion * 0.045
    + (light > 150) * 0.025
    + np.random.normal(0, 0.008, NUM_RECORDS)
)

# Значення споживання не може бути від'ємним.
energy = np.clip(energy, 0.005, None)

# ------------------------------------------------------------
# 2.5. Формування випадкових пропусків і викидів
# ------------------------------------------------------------

sensor_values = {
    1: light.copy(),
    2: energy.copy(),
    3: temperature.copy(),
    4: motion.astype(float)
}

sensor_names = {
    1: "Освітленість",
    2: "Споживання електроенергії",
    3: "Температура",
    4: "Рух"
}

sensor_units = {
    1: "lux",
    2: "kWh",
    3: "°C",
    4: "0/1"
}

records = []

for sensor_id, values in sensor_values.items():

    values = values.copy()

    # Додаємо випадкові пропущені значення.
    missing_mask = np.random.random(NUM_RECORDS) < 0.05
    values[missing_mask] = np.nan

    # Для аналогових сенсорів додаємо поодинокі викиди.
    if sensor_id in [1, 2, 3]:

        outlier_indices = np.random.choice(
            NUM_RECORDS,
            size=15,
            replace=False
        )

        for index in outlier_indices:
            if sensor_id == 1:
                values[index] += 500
            elif sensor_id == 2:
                values[index] += 0.5
            elif sensor_id == 3:
                values[index] += 15

    for timestamp, value in zip(timestamps, values):

        records.append({
            "timestamp": timestamp,
            "sensor_id": sensor_id,
            "sensor_name": sensor_names[sensor_id],
            "value": None if pd.isna(value) else float(value),
            "unit": sensor_units[sensor_id]
        })

df_raw = pd.DataFrame(records)

df_raw = df_raw.sort_values(
    ["timestamp", "sensor_id"]
).reset_index(drop=True)

# Зберігаємо початкові дані.
df_raw.to_csv(
    "data/raw_data.csv",
    index=False
)

print("\nЕТАП 1. ГЕНЕРАЦІЯ ДАНИХ")
print("Кількість сенсорів:", NUM_SENSORS)
print("Кількість часових точок:", NUM_RECORDS)
print("Загальна кількість записів:", len(df_raw))
print("Період моделювання:", df_raw["timestamp"].min(),
      "-", df_raw["timestamp"].max())
print("Кількість пропущених значень:", df_raw["value"].isna().sum())

display(df_raw.head(10))


# ============================================================
# 3. ПОПЕРЕДНЯ ОБРОБКА ДАНИХ НА РІВНІ EDGE COMPUTING
# ============================================================

def edge_preprocess(df):

    df = df.copy()

    df["timestamp"] = pd.to_datetime(df["timestamp"])

    processed_frames = []

    for sensor_id, group in df.groupby("sensor_id"):

        group = group.sort_values("timestamp").copy()
        group = group.set_index("timestamp")

        original_values = group["value"].copy()

        # Для датчика руху використовується дискретна обробка.
        if sensor_id == 4:

            group["value_clean"] = (
                group["value"].ffill().bfill().fillna(0)
            )

            group["value_clean"] = (
                group["value_clean"].round().clip(0, 1)
            )

            group["is_outlier"] = False
            group["value_norm"] = np.nan

        else:

            # Інтерполяція пропущених значень.
            group["value_interpolated"] = (
                group["value"].interpolate(
                    method="linear",
                    limit=10,
                    limit_direction="both"
                )
            )

            group["value_interpolated"] = (
                group["value_interpolated"].ffill().bfill()
            )

            # Виявлення викидів за допомогою медіани
            # та середнього абсолютного відхилення.
            rolling_median = (
                group["value_interpolated"]
                .rolling(window=11, center=True, min_periods=1)
                .median()
            )

            absolute_deviation = (
                group["value_interpolated"] - rolling_median
            ).abs()

            rolling_mad = (
                absolute_deviation
                .rolling(window=11, center=True, min_periods=1)
                .median()
            )

            threshold = np.maximum(
                4.5 * 1.4826 * rolling_mad,
                0.05 if sensor_id == 2 else 1.0
            )

            group["is_outlier"] = (
                absolute_deviation > threshold
            )

            # Заміна виявлених викидів локальною медіаною.
            group["value_clean"] = (
                group["value_interpolated"].copy()
            )

            group.loc[
                group["is_outlier"],
                "value_clean"
            ] = rolling_median[group["is_outlier"]]

            # Нормалізація очищених значень.
            mean_value = group["value_clean"].mean()
            std_value = group["value_clean"].std()

            if pd.isna(std_value) or std_value == 0:
                std_value = 1.0

            group["value_norm"] = (
                group["value_clean"] - mean_value
            ) / std_value

        # Повертаємо ідентифікатор сенсора.
        group["sensor_id"] = sensor_id

        group = group.reset_index()

        processed_frames.append(group)

    result = pd.concat(
        processed_frames,
        ignore_index=True
    )

    result = result.sort_values(
        ["timestamp", "sensor_id"]
    ).reset_index(drop=True)

    return result


print("\nЕТАП 2. EDGE COMPUTING")

start_processing = time.perf_counter()

df_processed = edge_preprocess(df_raw)

edge_time = time.perf_counter() - start_processing

# Зберігаємо результати обробки.
df_processed.to_csv(
    "data/processed_data.csv",
    index=False
)

print("Кількість записів після обробки:", len(df_processed))
print("Час обробки:", round(edge_time, 4), "сек.")
print("Кількість виявлених викидів:",
      int(df_processed["is_outlier"].sum()))
print("Кількість пропусків після обробки:",
      int(df_processed["value_clean"].isna().sum()))

display(
    df_processed[
        ["timestamp", "sensor_id", "value",
         "value_clean", "value_norm", "is_outlier"]
    ].head(10)
)


# ============================================================
# 4. СТВОРЕННЯ ЛОКАЛЬНОЇ РЕЛЯЦІЙНОЇ БАЗИ SQLITE
# ============================================================

print("\nЕТАП 3. СТВОРЕННЯ БАЗИ SQLITE")

DB_PATH = "iot_smart_home.sqlite"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)

# Активуємо контроль зовнішніх ключів.
conn.execute("PRAGMA foreign_keys = ON")

cursor = conn.cursor()

# ------------------------------------------------------------
# 4.1. Таблиця сенсорів
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE sensors (
    sensor_id INTEGER PRIMARY KEY,
    sensor_name TEXT NOT NULL,
    unit TEXT NOT NULL
)
""")

# ------------------------------------------------------------
# 4.2. Таблиця початкових вимірювань
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE raw_readings (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT NOT NULL,
    sensor_id INTEGER NOT NULL,
    value REAL,
    FOREIGN KEY(sensor_id) REFERENCES sensors(sensor_id)
)
""")

# ------------------------------------------------------------
# 4.3. Таблиця оброблених вимірювань
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE processed_readings (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT NOT NULL,
    sensor_id INTEGER NOT NULL,
    value_clean REAL,
    value_norm REAL,
    is_outlier INTEGER,
    FOREIGN KEY(sensor_id) REFERENCES sensors(sensor_id)
)
""")

# ------------------------------------------------------------
# 4.4. Таблиця подій системи
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE system_events (
    event_id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT NOT NULL,
    event_type TEXT NOT NULL,
    description TEXT
)
""")

# Додаємо інформацію про сенсори.
sensor_metadata = [
    (1, "Освітленість", "lux"),
    (2, "Споживання електроенергії", "kWh"),
    (3, "Температура", "°C"),
    (4, "Рух", "0/1")
]

cursor.executemany("""
INSERT INTO sensors(sensor_id, sensor_name, unit)
VALUES (?, ?, ?)
""", sensor_metadata)

# Підготовка початкових вимірювань.
raw_tuples = [
    (
        row.timestamp.isoformat(),
        int(row.sensor_id),
        None if pd.isna(row.value) else float(row.value)
    )
    for row in df_raw.itertuples(index=False)
]

cursor.executemany("""
INSERT INTO raw_readings(timestamp, sensor_id, value)
VALUES (?, ?, ?)
""", raw_tuples)

# Підготовка оброблених вимірювань.
processed_tuples = []

for row in df_processed.itertuples(index=False):

    processed_tuples.append((
        row.timestamp.isoformat(),
        int(row.sensor_id),
        float(row.value_clean),
        None if pd.isna(row.value_norm) else float(row.value_norm),
        int(bool(row.is_outlier))
    ))

cursor.executemany("""
INSERT INTO processed_readings(
    timestamp, sensor_id, value_clean, value_norm, is_outlier
)
VALUES (?, ?, ?, ?, ?)
""", processed_tuples)

# Створення індексів для прискорення запитів.
cursor.execute("""
CREATE INDEX idx_raw_timestamp
ON raw_readings(timestamp)
""")

cursor.execute("""
CREATE INDEX idx_processed_timestamp
ON processed_readings(timestamp)
""")

# Записуємо події системи.
events = [
    (
        START_TIME.isoformat(),
        "SYSTEM_START",
        "Запуск системи моніторингу розумного будинку"
    ),
    (
        START_TIME.isoformat(),
        "EDGE_PROCESSING",
        "Виконано попередню обробку даних на Edge-рівні"
    ),
    (
        datetime.utcnow().isoformat(),
        "DATABASE_SAVE",
        "Дані збережено у локальній базі SQLite"
    )
]

cursor.executemany("""
INSERT INTO system_events(timestamp, event_type, description)
VALUES (?, ?, ?)
""", events)

conn.commit()

# Перевірка структури бази.
print("Таблиці бази даних:")

tables = pd.read_sql_query("""
SELECT name FROM sqlite_master
WHERE type='table'
""", conn)

display(tables)

print("Кількість початкових записів:",
      pd.read_sql_query(
          "SELECT COUNT(*) AS total FROM raw_readings",
          conn
      ).iloc[0, 0])

print("Кількість оброблених записів:",
      pd.read_sql_query(
          "SELECT COUNT(*) AS total FROM processed_readings",
          conn
      ).iloc[0, 0])

conn.close()

print("Базу даних збережено:", DB_PATH)


# ============================================================
# 5. АНАЛІЗ ДАНИХ ТА ПОБУДОВА ГРАФІКІВ
# ============================================================

print("\nЕТАП 4. АНАЛІТИЧНА ОБРОБКА")

# Перетворюємо оброблені дані у зручний табличний формат.
df_wide = df_processed.pivot(
    index="timestamp",
    columns="sensor_id",
    values="value_clean"
).reset_index()

df_wide.columns = [
    "timestamp",
    "light_lux",
    "energy_kwh",
    "temperature_c",
    "motion"
]

df_wide["timestamp"] = pd.to_datetime(
    df_wide["timestamp"]
)

# Додаємо дату та годину.
df_wide["date"] = df_wide["timestamp"].dt.date
df_wide["hour"] = df_wide["timestamp"].dt.hour

# ------------------------------------------------------------
# 5.1. Добове споживання електроенергії
# ------------------------------------------------------------

daily_energy = (
    df_wide.groupby("date")["energy_kwh"]
    .sum()
    .reset_index()
)

daily_energy.columns = ["date", "energy_kwh"]

print("\nДобове споживання електроенергії:")
display(daily_energy)

# Графік добового споживання.
plt.figure(figsize=(10, 5))

plt.bar(
    daily_energy["date"].astype(str),
    daily_energy["energy_kwh"]
)

plt.title("Добове споживання електроенергії")
plt.xlabel("Дата")
plt.ylabel("Споживання, kWh")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig("plots/daily_energy.png", dpi=150)
plt.show()

# ------------------------------------------------------------
# 5.2. Графік зміни температури
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    df_wide["timestamp"],
    df_wide["temperature_c"]
)

plt.title("Зміна температури в приміщенні")
plt.xlabel("Час")
plt.ylabel("Температура, °C")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("plots/temperature.png", dpi=150)
plt.show()

# ------------------------------------------------------------
# 5.3. Графік рівня освітленості
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    df_wide["timestamp"],
    df_wide["light_lux"]
)

plt.title("Динаміка рівня освітленості")
plt.xlabel("Час")
plt.ylabel("Освітленість, lux")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("plots/lighting.png", dpi=150)
plt.show()

# ------------------------------------------------------------
# 5.4. Активність мешканців протягом доби
# ------------------------------------------------------------

hourly_activity = (
    df_wide.groupby("hour")["motion"]
    .mean()
    .reset_index()
)

hourly_activity["activity_percent"] = (
    hourly_activity["motion"] * 100
)

plt.figure(figsize=(10, 5))

plt.bar(
    hourly_activity["hour"],
    hourly_activity["activity_percent"]
)

plt.title("Активність мешканців протягом доби")
plt.xlabel("Година доби")
plt.ylabel("Активність, %")
plt.xticks(range(24))
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig("plots/activity.png", dpi=150)
plt.show()

print("\nСередня активність за годинами:")
display(
    hourly_activity[
        ["hour", "activity_percent"]
    ].round(2)
)

# ------------------------------------------------------------
# 5.5. Зв'язок споживання енергії з присутністю людей
# ------------------------------------------------------------

df_wide["occupancy_status"] = np.where(
    df_wide["motion"] >= 0.5,
    "Люди присутні",
    "Людей немає"
)

energy_by_occupancy = (
    df_wide.groupby("occupancy_status")["energy_kwh"]
    .mean()
    .reset_index()
)

print("\nСереднє споживання залежно від присутності:")
display(energy_by_occupancy)

plt.figure(figsize=(8, 5))

plt.bar(
    energy_by_occupancy["occupancy_status"],
    energy_by_occupancy["energy_kwh"]
)

plt.title("Споживання енергії залежно від присутності людей")
plt.xlabel("Стан приміщення")
plt.ylabel("Середнє споживання, kWh")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig("plots/energy_occupancy.png", dpi=150)
plt.show()

# ------------------------------------------------------------
# 5.6. Розрахунок показника енергоефективності
# ------------------------------------------------------------

total_energy = df_wide["energy_kwh"].sum()

# Корисне споживання - енергія, використана
# у часові інтервали, коли зафіксовано присутність людей.
useful_energy = df_wide.loc[
    df_wide["motion"] >= 0.5,
    "energy_kwh"
].sum()

idle_energy = total_energy - useful_energy

# Це умовний індикатор частки споживання
# в інтервалах присутності людей, а не нормативний ККД.
efficiency_coefficient = (
    useful_energy / total_energy * 100
    if total_energy > 0 else 0
)

print("\nПОКАЗНИКИ ЕНЕРГОЕФЕКТИВНОСТІ")
print("Загальне споживання:", round(total_energy, 3), "kWh")
print("Споживання за присутності людей:",
      round(useful_energy, 3), "kWh")
print("Споживання за відсутності людей:",
      round(idle_energy, 3), "kWh")
print("Умовний коефіцієнт енергоефективності:",
      round(efficiency_coefficient, 2), "%")

# ------------------------------------------------------------
# 5.7. Формування рекомендацій
# ------------------------------------------------------------

recommendations = []

idle_share = (
    idle_energy / total_energy * 100
    if total_energy > 0 else 0
)

if idle_share > 25:
    recommendations.append(
        "Рекомендується автоматично вимикати освітлення "
        "та непотрібні електропристрої за відсутності людей."
    )

if df_wide["light_lux"].mean() > 350:
    recommendations.append(
        "Доцільно використовувати природне освітлення "
        "та зменшувати роботу штучного освітлення вдень."
    )

if df_wide["temperature_c"].mean() > 23:
    recommendations.append(
        "Рекомендується перевірити налаштування опалення "
        "для уникнення надмірного нагрівання приміщення."
    )

if not recommendations:
    recommendations.append(
        "Поточні показники не виявили значних ознак "
        "неефективного використання електроенергії."
    )

print("\nРЕКОМЕНДАЦІЇ:")
for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")


# ============================================================
# 6. ПОРІВНЯННЯ АЛГОРИТМІВ СТИСНЕННЯ ДАНИХ
# ============================================================

print("\nЕТАП 5. ПОРІВНЯННЯ АЛГОРИТМІВ СТИСНЕННЯ")

# Для стиснення використовуємо оброблений набір даних.
csv_bytes = df_processed.to_csv(
    index=False
).encode("utf-8")

original_size = len(csv_bytes)

compression_results = {}

# ------------------------------------------------------------
# 6.1. GZIP
# ------------------------------------------------------------

start = time.perf_counter()
gzip_data = gzip.compress(csv_bytes, compresslevel=6)
gzip_time = time.perf_counter() - start

compression_results["gzip"] = {
    "data": gzip_data,
    "size": len(gzip_data),
    "time": gzip_time
}

# ------------------------------------------------------------
# 6.2. ZLIB
# ------------------------------------------------------------

start = time.perf_counter()
zlib_data = zlib.compress(csv_bytes, level=6)
zlib_time = time.perf_counter() - start

compression_results["zlib"] = {
    "data": zlib_data,
    "size": len(zlib_data),
    "time": zlib_time
}

# ------------------------------------------------------------
# 6.3. BZIP2
# ------------------------------------------------------------

start = time.perf_counter()
bz2_data = bz2.compress(csv_bytes, compresslevel=9)
bz2_time = time.perf_counter() - start

compression_results["bzip2"] = {
    "data": bz2_data,
    "size": len(bz2_data),
    "time": bz2_time
}

# ------------------------------------------------------------
# 6.4. Порівняння результатів
# ------------------------------------------------------------

compression_table = []

for method, result in compression_results.items():

    compressed_size = result["size"]

    compression_ratio = (
        compressed_size / original_size * 100
    )

    saved_percent = (
        100 - compression_ratio
    )

    compression_table.append({
        "Метод": method,
        "Початковий розмір, байт": original_size,
        "Стиснутий розмір, байт": compressed_size,
        "Економія, %": saved_percent,
        "Час стиснення, сек.": result["time"]
    })

compression_df = pd.DataFrame(compression_table)

display(
    compression_df.round(4)
)

# Визначаємо найкращий метод за розміром.
best_method = min(
    compression_results,
    key=lambda method: compression_results[method]["size"]
)

best_bytes = compression_results[best_method]["data"]

print("Найкращий метод за розміром:", best_method)
print("Розмір початкового набору:", original_size, "байт")
print("Розмір стиснутого набору:", len(best_bytes), "байт")

# Графік порівняння.
plt.figure(figsize=(8, 5))

plt.bar(
    compression_df["Метод"],
    compression_df["Стиснутий розмір, байт"]
)

plt.title("Порівняння розмірів стиснутих даних")
plt.xlabel("Алгоритм")
plt.ylabel("Розмір, байт")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig("plots/compression_comparison.png", dpi=150)
plt.show()

# Перевіряємо правильність розпакування.
if best_method == "gzip":
    restored_bytes = gzip.decompress(best_bytes)
elif best_method == "zlib":
    restored_bytes = zlib.decompress(best_bytes)
else:
    restored_bytes = bz2.decompress(best_bytes)

assert restored_bytes == csv_bytes

print("Перевірка розпакування пройдена успішно.")


# ============================================================
# 7. ПІДГОТОВКА ПІДСУМКОВОГО СТАТИСТИЧНОГО ЗВІТУ
# ============================================================

print("\nЕТАП 6. ФОРМУВАННЯ СТАТИСТИЧНОГО ЗВІТУ")

summary = {
    "project": "Smart Home IoT",
    "variant": 9,
    "records": int(len(df_raw)),
    "sensors": int(NUM_SENSORS),
    "missing_values_before": int(df_raw["value"].isna().sum()),
    "missing_values_after": int(
        df_processed["value_clean"].isna().sum()
    ),
    "outliers_detected": int(
        df_processed["is_outlier"].sum()
    ),
    "total_energy_kwh": round(float(total_energy), 3),
    "useful_energy_kwh": round(float(useful_energy), 3),
    "idle_energy_kwh": round(float(idle_energy), 3),
    "idle_energy_percent": round(float(idle_share), 2),
    "efficiency_indicator_percent": round(
        float(efficiency_coefficient), 2
    ),
    "average_temperature": round(
        float(df_wide["temperature_c"].mean()), 2
    ),
    "average_light": round(
        float(df_wide["light_lux"].mean()), 2
    ),
    "average_activity_percent": round(
        float(df_wide["motion"].mean() * 100), 2
    ),
    "best_compression": best_method,
    "original_size_bytes": int(original_size),
    "compressed_size_bytes": int(len(best_bytes)),
    "edge_processing_time": round(float(edge_time), 4),
    "recommendations": recommendations
}

# Записуємо звіт у JSON.
with open(
    "reports/final_report.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        ensure_ascii=False,
        indent=4
    )

print("\nПІДСУМКОВИЙ ЗВІТ")
print(json.dumps(
    summary,
    ensure_ascii=False,
    indent=4
))


# ============================================================
# 8. ІНТЕГРАЦІЯ З FIREBASE CLOUD STORAGE
# ============================================================

print("\nЕТАП 7. ІНТЕГРАЦІЯ З FIREBASE")

# Firebase використовується для хмарного зберігання
# оптимізованого набору даних.
#
# Для виконання цього етапу необхідні:
# 1. Проєкт Firebase.
# 2. Увімкнений Cloud Storage.
# 3. Service Account JSON.
# 4. Назва Storage Bucket.

firebase_success = False
firebase_downloaded = False

try:

    import firebase_admin
    from firebase_admin import credentials, storage

    from google.colab import files

    SERVICE_ACCOUNT = "serviceAccountKey.json"

    # Якщо ключ відсутній, пропонуємо завантажити його.
    if not os.path.exists(SERVICE_ACCOUNT):

        print("Завантажте файл Service Account JSON.")
        uploaded = files.upload()

        if SERVICE_ACCOUNT not in uploaded:
            raise FileNotFoundError(
                "Файл serviceAccountKey.json не завантажено."
            )

    # Вводимо назву Firebase Storage Bucket.
    FIREBASE_BUCKET = input(
        "Введіть назву Firebase Storage Bucket: "
    ).strip()

    if not FIREBASE_BUCKET:
        raise ValueError("Не вказано назву Firebase Bucket.")

    # Ініціалізація Firebase.
    if not firebase_admin._apps:

        cred = credentials.Certificate(SERVICE_ACCOUNT)

        firebase_admin.initialize_app(
            cred,
            {
                "storageBucket": FIREBASE_BUCKET
            }
        )

    bucket = storage.bucket()

    # Завантажуємо стиснутий набір даних.
    blob = bucket.blob(
        f"smart_home/{best_method}_processed_data.bin"
    )

    blob.upload_from_string(
        best_bytes,
        content_type="application/octet-stream"
    )

    print("Дані успішно завантажено у Firebase.")
    print("Шлях до файлу:", blob.name)

    firebase_success = True

    # --------------------------------------------------------
    # 8.1. Зворотне завантаження даних із Firebase
    # --------------------------------------------------------

    downloaded_bytes = blob.download_as_bytes()

    if best_method == "gzip":
        cloud_restored = gzip.decompress(downloaded_bytes)

    elif best_method == "zlib":
        cloud_restored = zlib.decompress(downloaded_bytes)

    else:
        cloud_restored = bz2.decompress(downloaded_bytes)

    # Відновлюємо CSV із хмарних даних.
    df_from_cloud = pd.read_csv(
        io.BytesIO(cloud_restored)
    )

    print("Дані успішно отримано з Firebase.")
    print("Кількість відновлених записів:",
          len(df_from_cloud))

    df_from_cloud.to_csv(
        "data/cloud_restored_data.csv",
        index=False
    )

    # Перевіряємо ідентичність даних.
    assert cloud_restored == csv_bytes

    print("Перевірка цілісності даних пройдена.")

    firebase_downloaded = True

    # --------------------------------------------------------
    # 8.2. Повторна статистична обробка даних
    # --------------------------------------------------------

    cloud_report = df_from_cloud.groupby(
        "sensor_id"
    )["value_clean"].agg(
        ["count", "mean", "std", "min", "max"]
    ).reset_index()

    print("\nСтатистичний звіт за даними з Firebase:")
    display(cloud_report.round(3))

    cloud_report.to_csv(
        "reports/cloud_statistics.csv",
        index=False
    )

except Exception as error:

    print("\nFirebase не завершив операцію.")
    print("Причина:", str(error))
    print(
        "Локальна частина проєкту виконана. "
        "Для хмарного етапу перевірте налаштування Firebase."
    )


# ============================================================
# 9. ФОРМУВАННЯ АРХІВУ РЕЗУЛЬТАТІВ ПРОЄКТУ
# ============================================================

print("\nЕТАП 8. ФОРМУВАННЯ АРХІВУ")

# Зберігаємо стиснутий набір даних окремим файлом.
with open(
    f"data/processed_data.{best_method}",
    "wb"
) as file:
    file.write(best_bytes)

# Формуємо архів з результатами.
ARCHIVE_PATH = "SmartHome_IoT_Variant9.zip"

with zipfile.ZipFile(
    ARCHIVE_PATH,
    "w",
    zipfile.ZIP_DEFLATED
) as archive:

    # Додаємо базу даних.
    archive.write(
        DB_PATH,
        arcname="database/iot_smart_home.sqlite"
    )

    # Додаємо початкові та оброблені дані.
    for filename in os.listdir("data"):

        path = os.path.join("data", filename)

        if os.path.isfile(path):
            archive.write(
                path,
                arcname=f"data/{filename}"
            )

    # Додаємо графіки.
    for filename in os.listdir("plots"):

        path = os.path.join("plots", filename)

        if os.path.isfile(path):
            archive.write(
                path,
                arcname=f"plots/{filename}"
            )

    # Додаємо звіти.
    for filename in os.listdir("reports"):

        path = os.path.join("reports", filename)

        if os.path.isfile(path):
            archive.write(
                path,
                arcname=f"reports/{filename}"
            )

print("Архів створено:", ARCHIVE_PATH)


# ============================================================
# 10. ПІДСУМКИ ВИКОНАННЯ ПРОГРАМИ
# ============================================================

print("\n" + "=" * 55)
print("ПРОГРАМУ УСПІШНО ЗАВЕРШЕНО")
print("=" * 55)

print("Варіант: 9")
print("Сенсорів:", NUM_SENSORS)
print("Загальна кількість записів:", len(df_raw))
print("Оброблених записів:", len(df_processed))
print("Edge Computing: виконано")
print("SQLite: виконано")
print("Аналітика та графіки: виконано")
print("Стиснення:", best_method)
print("Firebase завантаження:", firebase_success)
print("Firebase зворотне завантаження:", firebase_downloaded)
print("Архів:", ARCHIVE_PATH)

print("\nФайли результатів:")
print("- data/raw_data.csv")
print("- data/processed_data.csv")
print("- database:", DB_PATH)
print("- reports/final_report.json")
print("- plots/")
print("- архів:", ARCHIVE_PATH)

print("\nУсі локальні результати збережено.")


from google.colab import files

files.download("SmartHome_IoT_Variant9.zip")
```
